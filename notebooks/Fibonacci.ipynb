### Implementation 2 with Matrices

If we begin with the vector $\vec{u}=\langle 1,1\rangle$ and matrix $$A=\begin{pmatrix}0 & 1 \\ 0 & 1\end{pmatrix},$$ then, as we can see below, $A^n\vec{u}$ will give us the $(n^{th}$ $(n+1)^{st}$ Fibonacci numbers.

```python
u = np.array([1,1])
A = np.array([[0,1],[1,1]])
for i in range(5):
    print(f"Fibonacci number {i} is {u[0]} and Fibonacci number {i+1} is {u[1]}.")
    u = A@u
```

Using the numpy command `numpy.linalg.eig(a)` we can find approximations of the eigenvectors and values for $A$.

```python
eigvalues, eigvectors = LA.eig(A)

print(f"The eigenvalues of A are {eigvalues[0]} and {eigvalues[1]} with corresponding vectors \n\t {eigvectors[0]} \n and \n\t {eigvectors[1]}.")
```

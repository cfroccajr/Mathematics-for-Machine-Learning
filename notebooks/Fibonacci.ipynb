# Fibonacci Variation

The Fibonacci Sequence was originally a word problem about rabbits:
> "Suppose that a single pair of rabbits, one male and one female, are placed in a large enclosed garden with plenty of food.  In the first month juvenial rabbits do not reproduce, but in each subsequent month each pair of adult rabbits produce exactly one pair of juvenial rabbits.  How many rabbits are there in any given month?  How many rabbits are there after 12 months''

The Fibonacci sequence is the solution to this because at the start of month 1 there is one pair, at the start of month 2 there is still only one pair since juvenial rabbits don't reproduce, but at the start of month 3 there are two pair because the rabbits have aged enough to reproduce.  In general the number of pairs in month $n$ is equal to all the pairs in the previous month plus the number of rabbits from two months ago who are now old enough to have offspring, $$F_n=F_{n-1}+F_{n-2}.$$
